```python
import kagglehub

# Download dataset
path = kagglehub.dataset_download("jockeroika/ecommerce-data")

print("Data tersimpan di:", path)
```

```python
import pandas as pd
import os

# 1. Ambil nama file CSV yang ada di folder 'path' tadi
file_name = os.listdir(path)[0]
full_path = os.path.join(path, file_name)

# 2. Baca file tersebut sebagai tabel (DataFrame)
df = pd.read_csv(full_path)

# 3. Tampilkan datanya
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Supaya grafik muncul dengan bagus di Colab
%matplotlib inline
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Category', y='TotalAmount', data=df, estimator=sum, errorbar=None, palette='magma')
plt.title('Total Penjualan per Kategori Produk')
plt.xticks(rotation=45)
plt.show()
```

```python
# Ubah kolom OrderDate jadi format tanggal
df['OrderDate'] = pd.to_datetime(df['OrderDate'])

# Buat grafik garis
plt.figure(figsize=(12, 6))
df.groupby('OrderDate')['TotalAmount'].sum().plot(kind='line', marker='o', color='b')
plt.title('Tren Penjualan Harian')
plt.ylabel('Total Penjualan')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 8))
df['Rating'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=sns.color_palette('pastel'))
plt.title('Distribusi Rating Produk')
plt.ylabel('')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
# Membuat grafik batang total penjualan per kategori
sns.barplot(x='Category', y='TotalAmount', data=df, estimator=sum, palette='viridis')
plt.title('Total Revenue per Category')
plt.xticks(rotation=45)
plt.show()
```
